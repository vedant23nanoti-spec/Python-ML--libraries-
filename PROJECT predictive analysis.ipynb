```python
# 1 importing necessary packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#3 DATA CLEANING AND FILLING MISSING VALUES
df.isnull()
```

```python
df['Gender'].value_counts()
```

```python
df.Gender=df.Gender.fillna('Male')
```

```python
df['Married'].value_counts()
```

```python
df.Married=df.Married.fillna('Yes')
```

```python
df['Dependents'].value_counts()
```

```python
df.Dependents=df.Dependents.fillna('0')
```

```python
df['Self_Employed'].value_counts()
```

```python
df.Self_Employed=df.Self_Employed.fillna('0')
```

```python
df['Loan_Amount_Term'].value_counts()
```

```python
mean=df.LoanAmount.mean()
mean
```

```python
df.LoanAmount=df.LoanAmount.fillna(mean)
```

```python
df['Loan_Amount_Term'].value_counts()
```

```python
df.Loan_Amount_Term=df.Loan_Amount_Term.fillna(360.0)
```

```python
df['Credit_History'].value_counts()
```

```python
df.Credit_History=df.Credit_History.fillna(1.0)
```

```python
total=df.isnull().sum()
total
```

```python
df.head(30)
```

```python
#4 CONVERTING CATEGORICAL VARIABLE INTO A NUMERIC BY LABEL ENCODING
from sklearn.preprocessing import LabelEncoder
var_mod=['Gender','Married','Dependents','Education','Self_Employed','Property_Area','Loan_Status']
le=LabelEncoder()
for i in var_mod:
    df[i]=le.fit_transform(df[i])
    
```

```python
# 5 FEATURE SELECTION
#iloc is using for integer location based on idexing 
X=df.iloc[:,1:12]
y=df.iloc[:,12]
```

```python
X.head(10)
```

```python
y
```

```python
#6 SPLITTING DATA
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test= train_test_split (X,y,test_size=0.20,random_state=0)
# 80 data train : 20 data test
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
# 7 BUILDING DECISION TREE
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
y_pred
```

```python
y_test
```

```python
#8 MODEL EVALUTION AND ACCURACY
from sklearn import metrics
cnf_matrix=metrics.confusion_matrix(y_test,y_pred)
cnf_matrix
```

```python
print("Accuracy:",metrics.accuracy_score(y_test,y_pred))
```

```python
#9 BUILDING LOGISTIC REGRESSION MODEL
from sklearn.linear_model import LogisticRegression
logistic_regression=LogisticRegression()
logistic_regression.fit(X_train,y_train)
y_pred=logistic_regression.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
 #MODEL EVALUTION AND ACCURACY
from sklearn import metrics
cnf_matrix=metrics.confusion_matrix(y_test,y_pred)
cnf_matrix
```

```python
#10 VISUALISING CONFUSION NMATRIX USING HEATMAP
import seaborn as sn
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sn.heatmap(cnf_matrix,annot=True)
plt.title('Confusion matrix')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.show()
```

```python
#fmt= string formatting code to use when adding annotation
sn.heatmap(cnf_matrix/np.sum(cnf_matrix),annot=True,fmt='.2%')
plt.title('Confusion matrix')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.show()
```
