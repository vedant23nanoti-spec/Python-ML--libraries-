```python
import pandas as pd
import numpy as np
#example of series in pandas
data=np.array (['a','b','c','d'])
s=pd.Series(data,index=[0,1,2,3])
print(s)
 
```

```python
#example of dataframe
data1=[1,2,3,4,5]
df=pd.DataFrame(data1)
print(df)
```

```python
#2nd example
data2=[['Alex',10],['Bob',12],['Clark',13]]
df1=pd.DataFrame(data2,columns=['Name','Age'])
print (df1)
```

```python
#3rd example
data3={'Name':['tom','jack','steve','ricky'],'Age':[28,34,29,42]}
df2=pd.DataFrame(data3)
print (df2)
```
