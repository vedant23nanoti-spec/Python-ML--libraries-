```python
#regresion analysis
#simple linear regression

import numpy as np
from sklearn.linear_model import LinearRegression
```

```python

x= np.array([150,155,160,157,162,165,168,170,140,145,175,178,180]).reshape((-1,1))
y= np.array([50,52,54,58,62,63,66,68,45,47,70,72,73])
```

```python
x
```

```python
y
```

```python
x.shape
```

```python
y.shape
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show
```

```python
model= LinearRegression()
model.fit(x,y)
```

```python
print('intercept:',model.intercept_)
print('slope:',model.coef_)
```

```python
y_pred=model.predict(x)
print('predicted response:',y_pred)
```

```python
#lets find the weight of a person with height 166 cm
z=model.predict(np.array([[166]]))
print(z)
```

```python
y_pred=model.intercept_+ model.coef_ * x
print('predicted response:',y_pred)
```

```python
plt.scatter(x,y)
plt.plot(x,y_pred)
plt.show()
```

```python
# r value
r_sq=model.score(x,y)
print('coefficient of determination:',r_sq)

```
