```python
#regresion analysis
#simple linear regression

import numpy as np
from sklearn.linear_model import LinearRegression
```

```python

x= np.array([150,155,160,157,162,165,168,170,140,145,175,178,180]).reshape((-1,1))
y= np.array([50,52,54,58,62,63,66,68,45,47,70,72,73])
```

```python
x
```

```python
y
```

```python
x.shape
```

```python
y.shape
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show
```

```python
model= LinearRegression()
model.fit(x,y)
```

```python
print('intercept:',model.intercept_)
print('slope:',model.coef_)
```

```python
y_pred=model.predict(x)
print('predicted response:',y_pred)
```

```python
#lets find the weight of a person with height 166 cm
z=model.predict(np.array([[166]]))
print(z)
```

```python
y_pred=model.intercept_+ model.coef_ * x
print('predicted response:',y_pred)
```

```python
plt.scatter(x,y)
plt.plot(x,y_pred)
plt.show()
```

```python
# r value
r_sq=model.score(x,y)
print('coefficient of determination:',r_sq)

```

```python
#1IMPORT LIBRARIES
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

```

```python
#2 reading of data from csv 
customer_data=pd.read_csv("C:\\Users\\vedan\\Downloads\\datasets\\shopping_data.csv")
```

```python
customer_data.shape
```

```python
customer_data.head()
```

```python
#3 dataset has 5 columns but we have tom feature it in a 2d so only take 2 columns
data=customer_data.iloc[ : ,3: 5]
```

```python
data
```

```python
#4 need to know cluster and creating a dendogram
import scipy.cluster.hierarchy as shc
plt.title("Customer Dendogram")
dend=shc.dendrogram(shc.linkage(data,method='ward'))
```

```python
#5 to group the datapoints into 5 clusters
from sklearn.cluster import AgglomerativeClustering
cluster=AgglomerativeClustering(n_clusters=5,metric='euclidean',linkage='ward')
cluster.fit_predict(data)
```

```python
#6 we have 5 cluster we have 5 labels in output ie 0 to 4
```

```python
#7 lets plt cluster to see how many actually our data has clustered
plt.scatter(data.iloc[:,0],data.iloc[:,1] , c=cluster.labels_,cmap='rainbow')
```

```python
#2 import dataset
data=pd.read_csv("C:\\Users\\vedan\\Downloads\\datasets\\store_data.csv")
data.head()
```

```python
#3  DATA PREPROCCESING
transactions=[]
for i in range (0,7500): transactions.append([str(data.values[i,j]) for j in range(0,20)])
```

```python
transactions
```

```python
#4 training apriori model on dataset
from apyori import apriori
rules=apriori(transactions= transactions ,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)
```

```python
#5 DISPLAYING A RESULT OF RULES OCCURED FROM APRIORI FUNCTION
associations=list(rules)
associations
```
