```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.linear_model import LogisticRegression , LinearRegression
from sklearn.tree import DecisionTreeClassifier , DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier , RandomForestRegressor
from sklearn.svm import SVC , SVR
from sklearn.metrics import accuracy_score , mean_squared_error , r2_score , mean_absolute_error

```

```python
# Prepare features and target variables for classification and regression tasks

X=df.drop(columns=['timestamp', 'risk_classification', 'delivery_time_deviation'])
y_clf=df['risk_classification']
y_reg=df['delivery_time_deviation']
```

```python
#encoded categorical target variable for classification task
Le=LabelEncoder()
y_clf_encoded=Le.fit_transform(y_clf)
```

```python
#scale numerical features for regression task
scalar=StandardScaler()
X_scaled=scalar.fit_transform(X)
```

```python
#split the dataset into training and testing sets for classification task
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_scaled, y_clf_encoded, test_size=0.2, random_state=42)
```

```python
#split the dataset into training and testing sets for regression task
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_scaled, y_reg, test_size=0.2, random_state=42)
```

```python
#Classification Task
from sklearn.metrics import classification_report


print("---Classification Task---")
classification_models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree Classifier': DecisionTreeClassifier(random_state=42),
    'Random Forest Classifier': RandomForestClassifier(random_state=42),
    'Support Vector Classifier': SVC(random_state=42)
}

for model_name, model in classification_models.items():
    print(f"\nTraining {model_name}...")
    model.fit(X_train_clf, y_train_clf)
    y_pred_clf = model.predict(X_test_clf)
    accuracy = accuracy_score(y_test_clf, y_pred_clf)
    print(f"{model_name} Accuracy: {accuracy:.4f}")
    print(f"{model_name} Classification Report for {model_name}:\n{classification_report(y_test_clf, y_pred_clf, target_names=Le.classes_)}")
```
