```python
#Classification Task
from sklearn.metrics import classification_report


print("---Classification Task---")
classification_models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree Classifier': DecisionTreeClassifier(random_state=42),
    'Random Forest Classifier': RandomForestClassifier(random_state=42),
    'Support Vector Classifier': SVC(random_state=42)
}

for model_name, model in classification_models.items():
    print(f"\nTraining {model_name}...")
    model.fit(X_train_clf, y_train_clf)
    y_pred_clf = model.predict(X_test_clf)
    accuracy = accuracy_score(y_test_clf, y_pred_clf)
    print(f"{model_name} Accuracy: {accuracy:.4f}")
    print(f"{model_name} Classification Report for {model_name}:\n{classification_report(y_test_clf, y_pred_clf, target_names=Le.classes_)}")
```

```python
# COMPLETE CLASSIFICATION EVALUATION


from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
from sklearn.model_selection import cross_val_score
import pandas as pd
import matplotlib.pyplot as plt

print("\n" + "="*70)
print("COMPLETE CLASSIFICATION EVALUATION")
print("="*70)


# ------------------------------------------------------------
# 1. MODEL EVALUATION
# ------------------------------------------------------------

classification_results = []

for model_name, model in classification_models.items():

    print(f"\nEvaluating {model_name}...")

    # Train
    model.fit(X_train_clf, y_train_clf)

    # Predict
    y_pred_clf = model.predict(X_test_clf)

    # Metrics
    accuracy = accuracy_score(y_test_clf, y_pred_clf)

    precision = precision_score(
        y_test_clf,
        y_pred_clf,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test_clf,
        y_pred_clf,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test_clf,
        y_pred_clf,
        average="weighted",
        zero_division=0
    )

    classification_results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")

    print("\nClassification Report:")
    print(
        classification_report(
            y_test_clf,
            y_pred_clf,
            target_names=Le.classes_,
            zero_division=0
        )
    )


# ------------------------------------------------------------
# 2. MODEL COMPARISON TABLE
# ------------------------------------------------------------

classification_results_df = pd.DataFrame(
    classification_results
)

print("\n" + "="*70)
print("CLASSIFICATION MODEL COMPARISON")
print("="*70)

print(
    classification_results_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 3. 5-FOLD CROSS VALIDATION
# ------------------------------------------------------------

print("\n" + "="*70)
print("5-FOLD CROSS VALIDATION")
print("="*70)

for model_name, model in classification_models.items():

    cv_scores = cross_val_score(
        model,
        X_train_clf,
        y_train_clf,
        cv=5,
        scoring="accuracy"
    )

    print(
        f"{model_name}: "
        f"Mean Accuracy = {cv_scores.mean():.4f}, "
        f"Std = {cv_scores.std():.4f}"
    )


# ------------------------------------------------------------
# 4. CONFUSION MATRIX FOR EACH MODEL
# ------------------------------------------------------------

print("\n" + "="*70)
print("CONFUSION MATRICES")
print("="*70)

for model_name, model in classification_models.items():

    model.fit(X_train_clf, y_train_clf)

    y_pred_clf = model.predict(X_test_clf)

    cm = confusion_matrix(
        y_test_clf,
        y_pred_clf
    )

    print(f"\n{model_name}")
    print(cm)

    # Plot confusion matrix
    plt.figure(figsize=(6, 5))

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=Le.classes_
    )

    disp.plot(
        ax=plt.gca(),
        cmap="Blues",
        values_format="d"
    )

    plt.title(
        f"Confusion Matrix - {model_name}"
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 5. RANDOM FOREST FEATURE IMPORTANCE - FIXED
# ------------------------------------------------------------

print("\n" + "="*70)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("="*70)

rf_classifier = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(
    X_train_clf,
    y_train_clf
)

# Get feature names
if hasattr(X_train_clf, "columns"):
    feature_names = X_train_clf.columns.tolist()
else:
    feature_names = [
        f"Feature_{i+1}"
        for i in range(X_train_clf.shape[1])
    ]

# Create feature importance table
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_classifier.feature_importances_
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("\nFeature Importance:")
print(feature_importance.to_string(index=False))

# Plot top 15 features
top_features = feature_importance.head(15)

plt.figure(figsize=(9, 6))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 15 Feature Importances - Random Forest")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "="*70)
print("CLASSIFICATION EVALUATION COMPLETED")
print("="*70)

best_accuracy_model = classification_results_df.loc[
    classification_results_df["Accuracy"].idxmax(),
    "Model"
]

best_f1_model = classification_results_df.loc[
    classification_results_df["F1 Score"].idxmax(),
    "Model"
]

print(
    f"\nHighest Accuracy Model : {best_accuracy_model}"
)

print(
    f"Highest F1 Score Model : {best_f1_model}"
)

print("\nEvaluation includes:")
print("✓ Accuracy")
print("✓ Precision")
print("✓ Recall")
print("✓ F1 Score")
print("✓ Classification Report")
print("✓ 5-Fold Cross Validation")
print("✓ Confusion Matrix")
print("✓ Random Forest Feature Importance")
print("✓ Model Comparison")
```

```python
# REGRESSION TASK

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

print("\n---Regression Task---")

regression_models = {

    'Linear Regression': Pipeline([
        ('scaler', StandardScaler()),
        ('model', LinearRegression())
    ]),

    'Ridge Regression': Pipeline([
        ('scaler', StandardScaler()),
        ('model', Ridge(alpha=1.0))
    ]),

    'Decision Tree Regressor': DecisionTreeRegressor(
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    ),

    'Random Forest Regressor': RandomForestRegressor(
        n_estimators=300,
        max_depth=15,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    'Gradient Boosting Regressor': GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    'Support Vector Regressor': Pipeline([
        ('scaler', StandardScaler()),
        ('model', SVR(
            kernel='rbf',
            C=10,
            epsilon=0.1
        ))
    ])
}

for model_name, model in regression_models.items():

    print(f"\nTraining {model_name}...")

    model.fit(X_train_reg, y_train_reg)

    y_pred_reg = model.predict(X_test_reg)

    mse = mean_squared_error(y_test_reg, y_pred_reg)
    mae = mean_absolute_error(y_test_reg, y_pred_reg)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_reg, y_pred_reg)

    print(f"{model_name} MSE: {mse:.4f}")
    print(f"{model_name} MAE: {mae:.4f}")
    print(f"{model_name} RMSE: {rmse:.4f}")
    print(f"{model_name} R2 Score: {r2:.4f}")
```

```python
#Regression_Evaluation

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.model_selection import cross_val_score
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("\n" + "="*70)
print("COMPLETE REGRESSION EVALUATION")
print("="*70)

# ------------------------------------------------------------
# 1. MODEL EVALUATION
# ------------------------------------------------------------

regression_results = []

for model_name, model in regression_models.items():

    print(f"\nEvaluating {model_name}...")

    model.fit(X_train_reg, y_train_reg)

    y_pred_reg = model.predict(X_test_reg)

    mae = mean_absolute_error(y_test_reg, y_pred_reg)
    mse = mean_squared_error(y_test_reg, y_pred_reg)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_reg, y_pred_reg)

    regression_results.append({
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

    print(f"MAE      : {mae:.4f}")
    print(f"MSE      : {mse:.4f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"R2 Score : {r2:.4f}")


# ------------------------------------------------------------
# 2. MODEL COMPARISON
# ------------------------------------------------------------

regression_results_df = pd.DataFrame(
    regression_results
)

print("\n" + "="*70)
print("REGRESSION MODEL COMPARISON")
print("="*70)

print(
    regression_results_df.to_string(index=False)
)


# ------------------------------------------------------------
# 3. 5-FOLD CROSS VALIDATION
# ------------------------------------------------------------

print("\n" + "="*70)
print("5-FOLD CROSS VALIDATION (R2)")
print("="*70)

for model_name, model in regression_models.items():

    cv_scores = cross_val_score(
        model,
        X_train_reg,
        y_train_reg,
        cv=5,
        scoring="r2"
    )

    print(
        f"{model_name}: "
        f"Mean R2 = {cv_scores.mean():.4f}, "
        f"Std = {cv_scores.std():.4f}"
    )


# ------------------------------------------------------------
# 4. ACTUAL VS PREDICTED
# ------------------------------------------------------------

# Use Random Forest for visualization
rf_regressor = regression_models[
    "Random Forest Regressor"
]

rf_regressor.fit(
    X_train_reg,
    y_train_reg
)

y_pred_rf = rf_regressor.predict(
    X_test_reg
)

plt.figure(figsize=(7, 5))

plt.scatter(
    y_test_reg,
    y_pred_rf,
    alpha=0.5
)

min_value = min(
    y_test_reg.min(),
    y_pred_rf.min()
)

max_value = max(
    y_test_reg.max(),
    y_pred_rf.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted - Random Forest")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. RESIDUAL ANALYSIS
# ------------------------------------------------------------

residuals = y_test_reg - y_pred_rf

plt.figure(figsize=(7, 5))

plt.scatter(
    y_pred_rf,
    residuals,
    alpha=0.5
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.title("Residual Analysis - Random Forest")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. BASELINE COMPARISON
# ------------------------------------------------------------

print("\n" + "="*70)
print("BASELINE COMPARISON")
print("="*70)

baseline_prediction = np.full(
    len(y_test_reg),
    y_train_reg.mean()
)

baseline_mae = mean_absolute_error(
    y_test_reg,
    baseline_prediction
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test_reg,
        baseline_prediction
    )
)

baseline_r2 = r2_score(
    y_test_reg,
    baseline_prediction
)

print(f"Baseline MAE  : {baseline_mae:.4f}")
print(f"Baseline RMSE : {baseline_rmse:.4f}")
print(f"Baseline R2   : {baseline_r2:.4f}")


# ------------------------------------------------------------
# 7. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "="*70)
print("REGRESSION EVALUATION COMPLETED")
print("="*70)

best_r2_model = regression_results_df.loc[
    regression_results_df["R2 Score"].idxmax(),
    "Model"
]

lowest_mae_model = regression_results_df.loc[
    regression_results_df["MAE"].idxmin(),
    "Model"
]

lowest_rmse_model = regression_results_df.loc[
    regression_results_df["RMSE"].idxmin(),
    "Model"
]

print("\nBest R2 Model       :", best_r2_model)
print("Lowest MAE Model   :", lowest_mae_model)
print("Lowest RMSE Model  :", lowest_rmse_model)

print("\nIMPORTANT:")
print("Do not deploy the regression model yet if R2 is")
print("near 0 or negative. First verify the target/features.")
```

```python
import streamlit as st
import pandas as pd
import numpy as np
import joblib


# ============================================================
# PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Supply Chain Risk Predictor",
    page_icon="🚚",
    layout="wide"
)


# ============================================================
# LOAD DEPLOYMENT FILES
# ============================================================

@st.cache_resource
def load_models():

    model = joblib.load(
        "classification_model.pkl"
    )

    scaler = joblib.load(
        "classification_scaler.pkl"
    )

    features = joblib.load(
        "classification_features.pkl"
    )

    label_encoder = joblib.load(
        "classification_label_encoder.pkl"
    )

    return model, scaler, features, label_encoder


model, scaler, features, label_encoder = load_models()


# ============================================================
# HEADER
# ============================================================

st.title("🚚 Supply Chain Risk Prediction System")

st.write(
    "Enter the operational and logistics parameters "
    "to predict the delivery risk classification."
)

st.divider()


# ============================================================
# INPUT SECTION
# ============================================================

st.subheader("📊 Input Parameters")

st.info(
    "Enter values using the same units and ranges as the "
    "dataset used during model training."
)


# ------------------------------------------------------------
# CREATE TWO COLUMNS
# ------------------------------------------------------------

col1, col2 = st.columns(2)


# ============================================================
# INPUTS
# ============================================================

with col1:

    vehicle_gps_latitude = st.number_input(
        "Vehicle GPS Latitude",
        value=38.0,
        format="%.6f"
    )

    vehicle_gps_longitude = st.number_input(
        "Vehicle GPS Longitude",
        value=-90.0,
        format="%.6f"
    )

    fuel_consumption_rate = st.number_input(
        "Fuel Consumption Rate",
        min_value=0.0,
        value=8.0
    )

    eta_variation_hours = st.number_input(
        "ETA Variation (Hours)",
        min_value=0.0,
        value=2.9
    )

    traffic_congestion_level = st.number_input(
        "Traffic Congestion Level",
        min_value=0.0,
        value=5.0
    )

    warehouse_inventory_level = st.number_input(
        "Warehouse Inventory Level",
        min_value=0.0,
        value=300.0
    )

    loading_unloading_time = st.number_input(
        "Loading / Unloading Time",
        min_value=0.0,
        value=2.3
    )

    handling_equipment_availability = st.number_input(
        "Handling Equipment Availability",
        min_value=0.0,
        value=0.30,
        format="%.4f"
    )

    order_fulfillment_status = st.number_input(
        "Order Fulfillment Status",
        min_value=0.0,
        value=0.60,
        format="%.4f"
    )

    weather_condition_severity = st.number_input(
        "Weather Condition Severity",
        min_value=0.0,
        value=0.50,
        format="%.4f"
    )

    port_congestion_level = st.number_input(
        "Port Congestion Level",
        min_value=0.0,
        value=7.0
    )

    shipping_costs = st.number_input(
        "Shipping Costs",
        min_value=0.0,
        value=460.0
    )


with col2:

    supplier_reliability_score = st.number_input(
        "Supplier Reliability Score",
        min_value=0.0,
        value=0.50,
        format="%.4f"
    )

    lead_time_days = st.number_input(
        "Lead Time (Days)",
        min_value=0.0,
        value=5.2
    )

    historical_demand = st.number_input(
        "Historical Demand",
        min_value=0.0,
        value=6000.0
    )

    iot_temperature = st.number_input(
        "IoT Temperature",
        value=0.0
    )

    cargo_condition_status = st.number_input(
        "Cargo Condition Status",
        min_value=0.0,
        value=0.30,
        format="%.4f"
    )

    route_risk_level = st.number_input(
        "Route Risk Level",
        min_value=0.0,
        value=7.0
    )

    customs_clearance_time = st.number_input(
        "Customs Clearance Time",
        min_value=0.0,
        value=2.3
    )

    driver_behavior_score = st.number_input(
        "Driver Behavior Score",
        min_value=0.0,
        value=0.50,
        format="%.4f"
    )

    fatigue_monitoring_score = st.number_input(
        "Fatigue Monitoring Score",
        min_value=0.0,
        value=0.60,
        format="%.4f"
    )

    disruption_likelihood_score = st.number_input(
        "Disruption Likelihood Score",
        min_value=0.0,
        value=0.80,
        format="%.4f"
    )

    delay_probability = st.number_input(
        "Delay Probability",
        min_value=0.0,
        max_value=1.0,
        value=0.70,
        format="%.4f"
    )


# ============================================================
# PREDICTION BUTTON
# ============================================================

st.divider()

predict_button = st.button(
    "🔮 Predict Delivery Risk",
    type="primary",
    use_container_width=True
)


# ============================================================
# PREDICTION
# ============================================================

if predict_button:

    # --------------------------------------------------------
    # CREATE INPUT DATAFRAME
    # --------------------------------------------------------

    input_data = pd.DataFrame([{

        "vehicle_gps_latitude":
            vehicle_gps_latitude,

        "vehicle_gps_longitude":
            vehicle_gps_longitude,

        "fuel_consumption_rate":
            fuel_consumption_rate,

        "eta_variation_hours":
            eta_variation_hours,

        "traffic_congestion_level":
            traffic_congestion_level,

        "warehouse_inventory_level":
            warehouse_inventory_level,

        "loading_unloading_time":
            loading_unloading_time,

        "handling_equipment_availability":
            handling_equipment_availability,

        "order_fulfillment_status":
            order_fulfillment_status,

        "weather_condition_severity":
            weather_condition_severity,

        "port_congestion_level":
            port_congestion_level,

        "shipping_costs":
            shipping_costs,

        "supplier_reliability_score":
            supplier_reliability_score,

        "lead_time_days":
            lead_time_days,

        "historical_demand":
            historical_demand,

        "iot_temperature":
            iot_temperature,

        "cargo_condition_status":
            cargo_condition_status,

        "route_risk_level":
            route_risk_level,

        "customs_clearance_time":
            customs_clearance_time,

        "driver_behavior_score":
            driver_behavior_score,

        "fatigue_monitoring_score":
            fatigue_monitoring_score,

        "disruption_likelihood_score":
            disruption_likelihood_score,

        "delay_probability":
            delay_probability
    }])


    # --------------------------------------------------------
    # FORCE EXACT TRAINING FEATURE ORDER
    # --------------------------------------------------------

    input_data = input_data[
        features
    ]


    # --------------------------------------------------------
    # SCALE INPUT
    # --------------------------------------------------------

    input_scaled = scaler.transform(
        input_data
    )


    # --------------------------------------------------------
    # PREDICT
    # --------------------------------------------------------

    prediction_encoded = model.predict(
        input_scaled
    )[0]


    # --------------------------------------------------------
    # DECODE CLASS
    # --------------------------------------------------------

    prediction = label_encoder.inverse_transform(
        [prediction_encoded]
    )[0]


    # --------------------------------------------------------
    # DISPLAY RESULT
    # --------------------------------------------------------

    st.divider()

    st.subheader("🎯 Prediction Result")

    if prediction == "High Risk":

        st.error(
            f"⚠️ Delivery Risk: {prediction}"
        )

    elif prediction == "Moderate Risk":

        st.warning(
            f"⚠️ Delivery Risk: {prediction}"
        )

    else:

        st.success(
            f"✅ Delivery Risk: {prediction}"
        )


    # --------------------------------------------------------
    # SHOW ENCODED VALUE FOR TECHNICAL VERIFICATION
    # --------------------------------------------------------

    with st.expander("Technical Prediction Details"):

        st.write(
            "Encoded Prediction:",
            prediction_encoded
        )

        st.write(
            "Decoded Prediction:",
            prediction
        )

        st.write(
            "Input Features:",
            input_data
        )
```
