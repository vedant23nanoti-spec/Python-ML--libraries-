```python
#reshaping 3x3 array into 2x2x3 array
arr=np.array ([[1,2,3,4],
              [5,2,4,2],
              [1,2,0,1]])
newarr=arr.reshape(2,2,3)
print("\n Orignal array:\n",arr)
print("\n Reshaped array:\n",newarr)
```

```python
#Flatten array 
arr=np.array ([[1,2,3],[4,5,6]])
flarr=arr.flatten()
print("\n Original array:\n",arr)
print("\n Flatened array:\n",flarr)
print(arr.ndim)
print(flarr.ndim)
```

```python
#python programme to demonstrate 
#binary operations in numpy
a=np.array ([[1,2],
             [3,4]])
b=np.array ([[4,3],
             [2,1]])

#add array
print("Array sum:\n",a+b)
#multiplication of array(element wise)
print("Array multiplication:\n",a*b)
#matrix multiplication
print("Matrix multiplication:\n",a.dot(b))
```
