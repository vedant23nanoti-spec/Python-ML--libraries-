```python
import numpy as np
speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]
x=np.mean(speed)
print(x)
y=np.median(speed)
print(y)

# 89.7692 is mean of array
# 87.0 is median of array
```

```python
#mode
from scipy import stats
speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]
x=stats.mode(speed)
print(x)
```

```python
# Quartiles 
#lower upper middel
#  q1   q3     q2
```

```python
data=[1,3,3,4,5,6,6,7,8,8]
# first quartile(q1)
Q1=np.median (data[:5])
print(Q1)
#second quartile (q2)
Q2=np.median (data)
print(Q2)
# third quartile (Q3)
Q3=np.median (data[5:])
print(Q3)
# interquartile range (IQR)
IQR= Q3-Q1
print(IQR)


```
